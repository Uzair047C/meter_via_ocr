```python
import os

# ==========================
# CHANGE THESE PATHS
# ==========================
images_folder = r"D:\PaddleOCR\ocr_app\dataset\images\test"
label_file = r"D:\PaddleOCR\ocr_app\dataset\labels\test_list.txt"

# ==========================

# Get all image filenames
image_names = set(os.listdir(images_folder))

new_lines = []
removed = 0

with open(label_file, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()

        if not line:
            continue

        # Split on any whitespace (space or tab)
        parts = line.split()

        if len(parts) < 2:
            print(f"Skipping invalid line: {line}")
            continue

        image_name = parts[0]

        if image_name in image_names:
            new_lines.append(line + "\n")
        else:
            print(f"Deleting label: {image_name}")
            removed += 1

# Overwrite the label file
with open(label_file, "w", encoding="utf-8") as f:
    f.writelines(new_lines)

print(f"\nDone!")
print(f"Removed {removed} label entries.")
print(f"Remaining labels: {len(new_lines)}")
```
