```python
import subprocess
import sys

cmd = [
    sys.executable,                    # Uses the current Python interpreter
    "tools/predict_rec.py",
    "--use_gpu=False",
    "--image_dir=./dataset/images/test/0642_09681427069_crop_0.png",
    "--rec_model_dir=./models/digit_only_rec",
    "--rec_char_dict_path=./dataset/digit_dict.txt",
    "--rec_image_shape=3,48,320",
]

result = subprocess.run(cmd, capture_output=True, text=True)

print("===== STDOUT =====")
print(result.stdout)

print("===== STDERR =====")
print(result.stderr)

if result.returncode != 0:
    print(f"\nProcess exited with code {result.returncode}")
```
